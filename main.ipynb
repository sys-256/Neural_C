```python
import numpy as np
import pandas as pd # TODO needed?
from matplotlib import pyplot as plt
```

```python
def read_idx(filename):
    with open(filename, 'rb') as f:
        # Read the magic number (first 4 bytes)
        magic = int.from_bytes(f.read(4), 'big')
        
        # Determine the number of dimensions
        if magic in (2051, 2049):  # 2051 for images, 2049 for labels
            num_dims = 3 if magic == 2051 else 1
            # Read the remaining dimensions
            dims = []
            while True:
                dim = int.from_bytes(f.read(4), 'big')
                dims.append(dim)
                if len(dims) == num_dims:
                    break
            
            # Read the data
            data = np.fromfile(f, dtype=np.uint8)
            
            # Reshape the data into the appropriate dimensions
            if magic == 2051:
                # Images have shape (num_images, height, width)
                return data.reshape(dims[0], dims[1], dims[2])
            elif magic == 2049:
                # Labels have shape (num_labels,)
                return data
        else:
            raise ValueError("Invalid IDX file")

train_images = read_idx("data/train-images-idx3-ubyte")
train_labels = read_idx("data/train-labels-idx1-ubyte")
test_images = read_idx("data/t10k-images-idx3-ubyte")
test_labels = read_idx("data/t10k-labels-idx1-ubyte")
```
